```python
#Task 1: Sentiment Analysis
from transformers import pipeline
classifiers=pipeline("sentiment-analysis")
result=classifiers("I was happy to visit my home country")
print(result)
```

```python
#Task 2: Text Generation
from transformers import pipeline
text_generator=pipeline("text-generation",model="distilbert/distilgpt2")
generated_text=text_generator("Today is a rainy day in London",truncation=True,num_return_sequences=2)
print(generated_text)
```

```python
#Task 3:Question_Answering
from transformers import pipeline
qa_model=pipeline(task="question-answering")
question="What is my job"
context="I am developing AI models with python"
print(qa_model(question=question,context=context))
```

```python
import arxiv
import pandas as pd
from transformers import pipeline
#Query arxiv for AI/ML papers
query="Artificial Intelligence or Machine Learning"
search=arxiv.Search(
    query=query,
    max_results=10,
    sort_by=arxiv.SortCriterion.SubmittedDate
)
```

```python
import arxiv

#Extract metadata from results
papers=[]
client = arxiv.Client() # Instantiate the client
for result in client.results(search): # Use client.results() with the search object
  papers.append({
      'published':result.published,
      'title':result.title,
      'abstract':result.summary,
      'categories':result.categories
})
```

```python
#Convert to DataFrame
df=pd.DataFrame(papers)
pd.set_option('display.max_colwidth', None)
```

```python
#Display the first 10 papers
df.head(10)
```

```python
#Example: take first abstract
abstract=df['abstract'][0]
```

```python
#Load summarization model
from transformers import pipeline
summarizer=pipeline("summarization",model="facebook/bart-large-cnn")
```

```python
#Summarize abstract
summarization_result=summarizer(abstract)
summarization_result[0]['summary_text']
```
